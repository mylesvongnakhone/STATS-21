### 1. Intercollegiate Math Tournament (Registration opens tonight!)
Write a function that takes in an integer input `n` and returns a string array where:
* `output[i] == "Intercollegiate Math Tournament"` if `i` is divisible by 8 and 5
* `output[i] == "Intercollegiate"` if `i` is divisible by 8
* `output[i] == "Math Tournament"` if `i` is divisble by 5
* `output[i] == i` as a string otherwise

```python
def icmt(n):
    result = []
    for i in range(1, n + 1):
        if i % 8 == 0 and i % 5 == 0:
            result.append("Intercollegiate Math Tournament")
        elif i % 8 == 0:
            result.append("Intercollegiate")
        elif i % 5 == 0:
            result.append("Math Tournament")
        else:
            result.append(str(i))
    return result
```

### 2. Collatz Conjecture
Write a function that takes a positive integer `n` and if `n` is
* even: divide it by 2
* odd: multiply by 3 and add 1

The function should return the number of steps that it takes for `n` |to reach 1

```python
def collatz(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
        steps += 1
    return steps
```

```python
collatz(1) #0
collatz(2) #1
collatz(3) #7
collatz(6) #8
collatz(27) #111
```

### 3. Fibonacci Number
Write a function to calculate the nth Fibonacci Number

0, 1, 1, 2, 3, 5, 8, 13, 21, ...

where 0 is the zeroth Fibonacci number

```python
def fib(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a
```

```python
fib(0) #0
fib(1) #1
fib(2) #1
fib(5) #5
fib(10) #55
fib(45) #1134903170
fib(100) #354224848179261915075
```
