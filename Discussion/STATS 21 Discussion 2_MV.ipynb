### Write a functions that takes in 2 numbers and adds, subtracts, multiplies, and divides them respectively

````python
def add(x,y):
    return x + y

def subtract(x,y):
    return x - y

def multiply(x,y):
    return x * y

def divide(x,y):
    return x / y

x = 6
y = 7

print("x:", x)
print("y:", y)
print("add:", add(x,y))
print("subtract:", subtract(x,y))
print("multiply:", multiply(x,y))
print("divide:", divide(x,y))  
````
